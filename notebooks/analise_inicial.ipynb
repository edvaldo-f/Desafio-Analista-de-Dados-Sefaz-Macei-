```python
import pandas as pd
import pyarrow
import numpy as np
from pathlib import Path
```

```python
# Encontrando o arquivo.

raiz_do_projeto = Path.cwd().parent
caminho_db = raiz_do_projeto / "dados_otimizados" / "dados_otimizados.parquet"
```

### Fazendo leituras inicias
Conhecendo a base de dados...

```python
df = pd.read_parquet(caminho_db)

df.head()
```

```python
# Verificando valores nulos e duplicados

print("Número de valores nulos:\n", df.isnull().sum())
print("\nNúmero de valores duplicados:", df.duplicated().sum())
```

Contagem de dados em cada ano

```python
contagem = df['Ano'].value_counts()

print("Contagem de valores únicos na coluna 'Ano':\n", contagem)
```
