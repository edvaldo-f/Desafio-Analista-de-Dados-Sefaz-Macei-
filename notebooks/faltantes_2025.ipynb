# Quais as capitais faltantes em 2025?

Como informado no README, 2025 possui dados incompletos referentes as suas capitais, esse arquivo serve para indicar quais delas 2025 possui e quais não possui.

```python
import pandas as pd
from pathlib import Path

# Encontrando o arquivo
raiz_do_projeto = Path.cwd().parent
caminho_db = raiz_do_projeto / "dados_otimizados" / "dados_otimizados.parquet"

# Lendo o arquivo

df = pd.read_parquet(caminho_db)
```

```python
# Verificando através dos dados do ano de 2024

# Estados, para uma saída mais explicada, sem as siglas dos UF
estados_brasil = {
    'AC': 'Acre', 'AL': 'Alagoas', 'AP': 'Amapá', 'AM': 'Amazonas',
    'BA': 'Bahia', 'CE': 'Ceará', 'ES': 'Espírito Santo',
    'GO': 'Goiás', 'MA': 'Maranhão', 'MT': 'Mato Grosso', 'MS': 'Mato Grosso do Sul',
    'MG': 'Minas Gerais', 'PA': 'Pará', 'PB': 'Paraíba', 'PR': 'Paraná',
    'PE': 'Pernambuco', 'PI': 'Piauí', 'RJ': 'Rio de Janeiro', 'RN': 'Rio Grande do Norte',
    'RS': 'Rio Grande do Sul', 'RO': 'Rondônia', 'RR': 'Roraima', 'SC': 'Santa Catarina',
    'SP': 'São Paulo', 'SE': 'Sergipe', 'TO': 'Tocantins'
}

# Isolando valores únicos de 2025
uf_2025 = df[df["Ano"] == 2025]["UF"].unique()

# Filtrando o DataFrame para 2024
df_2024 = df[df["Ano"] == 2024]

# Coletando as UF que estão em 2024, mas não em 2025
uf_diferentes = df_2024[~df_2024["UF"].isin(uf_2025)]["UF"].unique()

# Convertendo as siglas
estados_diferentes = [estados_brasil[sigla] for sigla in uf_diferentes if sigla in estados_brasil]

print("Capitais faltantes em 2025: ")
for estados in estados_diferentes:
    print(estados)

```
